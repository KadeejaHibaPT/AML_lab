```python
# Install package (run once in Google Colab)
!pip -q install pgmpy

import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination

# Student dataset transcribed from the prompt
df = pd.DataFrame({
    "Study":       ["High", "High", "High", "Low", "Low", "Low",
                    "High", "High", "High", "Low", "Low", "Low"],
    "Attendance":  ["Good", "High", "Low", "Low", "Low", "Good",
                    "Good", "Good", "Poor", "Poor", "Poor", "Poor"],
    "Result":      ["High", "High", "Low", "Low", "Low", "Good",
                    "Good", "Poor", "Poor", "Poor", "Pass", "Pass"],
    "Scholarship": ["High", "Low", "Low", "Low", "Low", "Good",
                    "Good", "Poor", "Poor", "Poor", "Fail", "No"]
})

# The pasted table has unclear row alignment and inconsistent category values.
# Replace the entries above with correctly aligned Study, Attendance,
# Result, and Scholarship values before interpreting the inference results.

# Define network structure
model = DiscreteBayesianNetwork([
    ("Study", "Result"),
    ("Attendance", "Result"),
    ("Result", "Scholarship")
])

# Learn CPDs using Maximum Likelihood Estimation
model.fit(df, estimator=MaximumLikelihoodEstimator)

# Display CPDs and validate the model
for cpd in model.get_cpds():
    print(cpd)

print("Model valid:", model.check_model())

# Visualize the network
nx.draw(
    model, with_labels=True, node_size=2500,
    node_color="lightblue", arrowsize=20
)
plt.show()

# Probabilistic inference using Variable Elimination
infer = VariableElimination(model)

print("\na) P(Result | Study=Low, Attendance=Poor)")
print(infer.query(
    ["Result"], evidence={"Study": "Low", "Attendance": "Poor"}
))

print("\nb) P(Scholarship | Result=Fail)")
print(infer.query(
    ["Scholarship"], evidence={"Result": "Fail"}
))

print("\nc) P(Study | Result=Pass)")
print(infer.query(
    ["Study"], evidence={"Result": "Pass"}
))

print("\nd) P(Result)")
print(infer.query(["Result"]))
```
