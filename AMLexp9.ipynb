```python
import numpy as np

# Hidden states and observations
states = ["Healthy", "Recovering"]
observations = ["J1", "J2", "J3"]

# Initial probabilities: P(state at time 1)
initial = np.array([0.45, 0.55])

# Transition matrix A:
# rows = current state, columns = next state
transition = np.array([
    [0.7, 0.3],  # Healthy -> Healthy, Recovering
    [0.5, 0.5]   # Recovering -> Healthy, Recovering
])

# Emission matrix B:
# rows = hidden state, columns = observation
emission = np.array([
    [0.8, 0.1, 0.1],  # Healthy emits J1, J2, J3
    [0.2, 0.3, 0.5]   # Recovering emits J1, J2, J3
])

# Observed jockey sequence
obs_sequence = ["J2", "J3", "J1"]
obs_indices = [observations.index(obs) for obs in obs_sequence]

# Viterbi tables
num_states = len(states)
num_steps = len(obs_sequence)
viterbi = np.zeros((num_states, num_steps))
backpointer = np.zeros((num_states, num_steps), dtype=int)

# Step 1: Initialization
viterbi[:, 0] = initial * emission[:, obs_indices[0]]

# Steps 2+: Recursion
for t in range(1, num_steps):
    for current_state in range(num_states):
        scores = (
            viterbi[:, t - 1]
            * transition[:, current_state]
        )
        best_previous_state = np.argmax(scores)
        viterbi[current_state, t] = (
            scores[best_previous_state]
            * emission[current_state, obs_indices[t]]
        )
        backpointer[current_state, t] = best_previous_state

# Step 3: Find the most likely final state
last_state = np.argmax(viterbi[:, -1])
best_path = [last_state]

# Step 4: Backtrack to recover the state sequence
for t in range(num_steps - 1, 0, -1):
    last_state = backpointer[last_state, t]
    best_path.append(last_state)

best_path.reverse()
decoded_states = [states[i] for i in best_path]
best_probability = np.max(viterbi[:, -1])

# Display results
print("Observation sequence:", " -> ".join(obs_sequence))
print("Most likely hidden states:", " -> ".join(decoded_states))
print(f"Path probability: {best_probability:.6f}")

print("\nViterbi probability table:")
print("Rows = states (Healthy, Recovering); columns = time steps")
print(viterbi)
```
